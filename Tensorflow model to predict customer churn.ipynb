```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
url = 'https://raw.githubusercontent.com/nicknochnack/Tensorflow-in-10-Minutes/main/Churn.csv'
df = pd.read_csv(url)
df.head()
```

```python
#Preprocessing stage by using one hot encoding to convert data to numerical to be used in modeling
#Assigning the x and y variable
x = pd.get_dummies(df.drop(['Churn', 'Customer ID'], axis=1))
y = df['Churn'].apply(lambda x:1 if x =='Yes' else 0)
```

```python
#Splitting training and test dataset
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=.2)
```

```python
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense
from sklearn.metrics import accuracy_score
```

```python
#Creating Sequential model, with linear stack of layers that can be used to build neural networks in a straightforward and simple manner
model = Sequential()
#Adding hidden layers
#In the parameters, input will be the same amt of dimension available in x_train column
model.add(Dense(units=32, activation='relu', input_dim=len(x_train.columns)))
model.add(Dense(units=64, activation='relu'))
#Output layer with 1, to predict yes or no result
model.add(Dense(units=1, activation='sigmoid'))
```

```python
#Compiling model by ordering the model to use specific optimizer, metrics in the parameters
model.compile(loss='binary_crossentropy', optimizer='sgd', metrics=['accuracy'])
```

```python
#Training the model with training sets split earlier
model.fit(x_train, y_train, epochs=200, batch_size=32)
```

```python
#Using the model for Predictions on the test sets
y_pred = model.predict(x_test)

#Convert result into binary outcome by using list comprehension
y_pred = [0 if val <0.5 else 1 for val in y_pred]
```

```python
#Evaluate results using accuracy score from sklearn
accuracy_score(y_test, y_pred)
```

```python
# Data Visualization on prediction results VS y_test
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
```

```python
# Generate confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)
```

```python
# Plot confusion matrix using Seaborn
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
# Generate classification report
class_report = classification_report(y_test, y_pred)
print("Classification Report:\n", class_report)
```

```python
# Visualize the distribution of predictions and actual values
plt.figure(figsize=(12, 6))
# Distribution of predicted values
plt.subplot(1, 2, 1)
sns.histplot(y_pred, bins=2, kde=False)
plt.title('Distribution of Predicted Values')
plt.xlabel('Predicted Value')
plt.ylabel('Frequency')

# Distribution of actual values
plt.subplot(1, 2, 2)
sns.histplot(y_test, bins=2, kde=False)
plt.title('Distribution of Actual Values')
plt.xlabel('Actual Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
