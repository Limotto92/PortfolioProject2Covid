```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/futurexskill/projects/main/knn-classification/purchase_history.csv"
    
df = pd.read_csv(url)
df
```

```python
df.count()
len(df)
```

```python
# Converting categorical values to numerical using One hot encoding
gender_encoded = pd.get_dummies(df['Gender'])
```

```python
gender_encoded
```

```python
# Reduce the variable (redundancy) to avoid overfeeding into the ML model, this 1 column containing 0, 1 includes both F and M in a single row
gender_encoded = pd.get_dummies(df['Gender'], drop_first=True)
```

```python
# Concat the value into original df
df = pd.concat([df, gender_encoded], axis=1)
```

```python
# Select relevant features for ML model and cast in numpy arrays
import numpy as np
```

```python
# Assigning variable x and y
x = df[['Male','Age','Salary','Price']].to_numpy()
x
```

```python
y = df['Purchased'].to_numpy()
y
```

```python
#Seperate the dataset into training set and test set to train and evaluate the ML model
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
x_train
```

```python
# Scaling features
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

```python
#Building ML model with KNN algorithm
from sklearn.neighbors import KNeighborsClassifier

k = 5
knn = KNeighborsClassifier(n_neighbors=k)
```

```python
# Training the knn model with training data split earlier
knn.fit(x_train, y_train)
```

```python
# Testing the model to do predictions
y_pred = knn.predict(x_test)
y_pred
```

```python
#Compare the predictions with the actual test data using accuracy score method from scikitlearn libraby
from sklearn.metrics import accuracy_score
```

```python
# Accuracy comparison results projected in %
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
# Testing the model into new dataset of 10 new customers
df2 = pd.read_csv('https://raw.githubusercontent.com/futurexskill/projects/main/knn-classification/new_customers.csv')
df2
```

```python
# Peforming One Hot Encoding and all neccesary steps to standardise the data format as our 1st df to be able to be fitted in ML
gender_encoded_2 = pd.get_dummies(df2['Gender'], drop_first=True)
gender_encoded_2
```

```python
df2 = pd.concat([df2, gender_encoded_2], axis=1)
df2
```

```python
# Assigning x2, y2 variables
x2 = df2[['Male','Age','Salary','Price']].to_numpy()
x2
```

```python
x2_scale = scaler.fit_transform(x2)
x2_scale
```

```python
y2_pred = knn.predict(x2_scale)
y2_pred
```

```python
# Add the prediction table to the df
df2['will_purchase'] = y2_pred
df2
```
